## Crear External Location

```python
%sql
-- CREATE WIDGET
CREATE WIDGET TEXT storageName DEFAULT "adlsrenzocavero";

-- CREATE EXTERNAL LOCATIONS
CREATE EXTERNAL LOCATION IF NOT EXISTS `extl-metastore`
URL 'abfss://metastore@adlsrenzocavero.dfs.core.windows.net/'
WITH (STORAGE CREDENTIAL `credential-renzocavero`);

CREATE EXTERNAL LOCATION IF NOT EXISTS `extl-raw`
URL 'abfss://raw@adlsrenzocavero.dfs.core.windows.net/'
WITH (STORAGE CREDENTIAL `credential-renzocavero`);

CREATE EXTERNAL LOCATION IF NOT EXISTS `extl-bronze`
URL 'abfss://bronze@adlsrenzocavero.dfs.core.windows.net/'
WITH (STORAGE CREDENTIAL `credential-renzocavero`);

CREATE EXTERNAL LOCATION IF NOT EXISTS `extl-silver`
URL 'abfss://silver@adlsrenzocavero.dfs.core.windows.net/'
WITH (STORAGE CREDENTIAL `credential-renzocavero`);

CREATE EXTERNAL LOCATION IF NOT EXISTS `extl-golden`
URL 'abfss://golden@adlsrenzocavero.dfs.core.windows.net/'
WITH (STORAGE CREDENTIAL `credential-renzocavero`);
```

## Crear Catalog & Schema

```python
%sql
CREATE CATALOG IF NOT EXISTS catalog_dev
MANAGED LOCATION 'abfss://metastore@adlsrenzocavero.dfs.core.windows.net/catalog_dev/';

-- BRONZE
CREATE SCHEMA IF NOT EXISTS catalog_dev.bronze
MANAGED LOCATION 'abfss://bronze@adlsrenzocavero.dfs.core.windows.net/';

-- SILVER
CREATE SCHEMA IF NOT EXISTS catalog_dev.silver
MANAGED LOCATION 'abfss://silver@adlsrenzocavero.dfs.core.windows.net/';

-- GOLD
CREATE SCHEMA IF NOT EXISTS catalog_dev.gold
MANAGED LOCATION 'abfss://gold@adlsrenzocavero.dfs.core.windows.net/';
```
